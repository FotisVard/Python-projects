```python
import matplotlib.pyplot as plt
import scipy as sp
import math
import numpy as np
```

```python
r = range(0,5)
omegae = 1
h = 2.08366191 * 10**10
hbar = h/(2*math.pi)

def Energy(r):
    return hbar * omegae * (r + 0.5)

for r in range(0,5):
    print("Energy level for r =", r, "is", Energy(r))
```

```python
plt.plot(range(0,5), [Energy(r) for r in range(0,5)], marker='o')
plt.title("Energy Levels of a Quantum Harmonic Oscillator (Einstein Model)")
plt.xlabel("Quantum Number (r)")
plt.ylabel("Energy (Joules)")
plt.xticks(range(0,5))
plt.grid()
plt.show()
```
